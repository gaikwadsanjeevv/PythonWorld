```python

def find_largest(numbers):
    largest = numbers[0]

    for i in range(1, len(numbers)):
        if numbers[i] > largest:
            largest = numbers[i]

    return largest


numbers = [12, 7, 25, 9, 18]

print(find_largest(numbers))
```

Step 2: Develop the thinking process
How would you solve this problem manually?

You would remember the first number, compare it with the next number and update your remembered value whenever you find a larger number.

You would repeat this process until all numbers have been checked.

This is the foundation of an algorithm called linear traversal.

Step 3: Write the algorithm
1. Assume the first number is the largest.
2. Traverse the remaining elements.
3. Compare each element with the current largest number.
4. If the current element is larger, update the largest number.
5. Return the largest number after completing the traversal.

Write Pseudocode:  

START

SET largest = first element

FOR each remaining element:
    IF element > largest:
        SET largest = element

RETURN largest

END

```python
# String slicing: Slicing allows us to extract a portion of a string (stop index is exclusive).
str = "Hello, World!"
print(str[0:5])       # Output: Hello | Start at index 0 up to (excluding) 5, step +1.
print(str[7:])        # Output: World! | Start at index 7 to the end of the string, step +1.
print(str[:5])        # Output: Hello | Start at index 0 (default) up to (excluding) 5, step +1.
print(str[::2])       # Output: Hlo ol! | Start at 0 to end, taking every 2nd character (step +2).
print(str[::-1])      # Output: !dlroW ,olleH | Reverse the entire string (step -1 from end to start).
print(str[-6:-1])      # Output: World | Start at -6 ('W') up to (excluding) -1 ('!'), step +1.
print(str[-1:-6:-1])   # Output: !dlro | Start at -1 ('!') moving backward up to (excluding) -6 ('W'), step -1.
print(str[-1:-6:-2])   # Output: !lo | Start at -1 ('!') moving backward up to (excluding) -6 ('W'), step -2.
print(str[-1:-6:3])    # Output:  | Empty string (""): positive step (+3) moves forward, so -1 to -6 is empty.
```
